```python
from deepface import DeepFace
import cv2
import numpy as np
import time

MIN_ADULT_AGE = 18
CAPTURE_TIME = 3   # seconds

def verify_age():
    cap = cv2.VideoCapture(0)
    ages = []
    start_time = time.time()

    while time.time() - start_time < CAPTURE_TIME:
        ret, frame = cap.read()
        if not ret:
            continue

        try:
            result = DeepFace.analyze(
                frame,
                actions=["age"],
                enforce_detection=False
            )
            age = result[0]["age"]
            ages.append(age)

            label = f"Detecting age... {age}"
            cv2.putText(frame, label, (20, 40),
                        cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 0), 2)

        except:
            pass

        cv2.imshow("Age Verification", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()

    if len(ages) == 0:
        return None   # no face

    avg_age = int(np.mean(ages))
    print(f"📊 Average detected age: {avg_age}")

    if avg_age < MIN_ADULT_AGE:
        return "child"
    else:
        return "adult"


print("====== AGE BASED APPLICATION LAUNCHER ======")
print("1. Open Browser")
print("2. Open Game (Example)")
print("3. Open Media Player")

choice = input("Enter your choice: ")

result = verify_age()

if result == "child":
    print("👶 Child detected → Opening Media Player only")
    os.system("start wmplayer")

elif result == "adult":
    print("🧑 Adult detected → Access granted")

    if choice == "1":
        os.system("start chrome")
    elif choice == "2":
        os.system("start notepad")
    elif choice == "3":
        os.system("start wmplayer")
    else:
        print("Invalid choice")

else:
    print("❌ No face detected")
```
